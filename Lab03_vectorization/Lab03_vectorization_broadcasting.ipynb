```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])

result = x ** 2

print(result)
```

```python
y = 2 *x + 5
print(y)
```

- **Vectorisation** is applying an operation to a whole array at once instead of element by element in a python loop.
- **Vectoorisation is useful because:** shorter, more readable code and much faster since the looping happens in optimized compiled C code.
- **Loop vs vectorized:** a python loop is interpreted one element at a time with overhead each step: a Numpy operation runs in a single compiled pass over contiguous memory.

```python
marks = np.array([40, 55, 62, 70, 81, 90])
adjusted = marks + 5
percentage = marks / 100

print("Adjusted Marks:", adjusted)
print("Percentages:", percentage)
```

```python
z = (marks - np.mean(marks)) / np.std(marks)
print(z.mean())
```

Broadcasting

```python

data = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [90, 85, 95]
])

adjustment = np.array([5, 0, 2])

result = data + adjustment

print("adjustment =", adjustment)
print("result =")
print(result)
```

- `data` is 3x3, `adjustment` is `(3,)`. NumPy **stretches** the smaller array virtually across the rows, so `[5, 0, 2]` is added to every row: column 1 gets +5, column 2 gets +0, column 3 gets +2. Rule: compare shapes from the right; dimensions are compatible if equal or one of them is 1.

```python
data = np.array([
[10, 20, 30],
[20, 30, 40],
[30, 40, 50]
])
means = data.mean(axis=0)
centered = data - means
print("means=", means)
print("centered=")
print(centered)
```

```python
print(centered.mean(axis=0))
```

- This works because `means` has shape `(3,)` which broadcasts across the 3 rows, so each column has its own mean subtracted.

```python
x = np.arange(1_000_000)
result_loop = []
for value in x:
    result_loop.append(value**2 + 2*value + 1)
result_vectorized = x**2 + 2*x + 1
np.allclose(result_loop, result_vectorized) 
```

1. Yes, `True`.
2. The vectorized one is easier to read (it looks like the maths).
3. The vectorized one is faster.
4. Because it avoids Python's per-element interpreter overhead and uses compiled, memory-efficient array operations.

```python
np.random.seed(123)
x = np.random.normal(100, 15, 100_000)

z = (x - x.mean()) / x.std()
print(z.mean())
print(z.std())
```

**Interpretation:** Standardised values have mean 0 and standard deviation 1. Vectorization matters for big datasets because loops over hundreds of thousands or millions of values are slow, while vectorized code processes them all in milliseconds with less code.

```python

def standardize(x):
    return (x - np.mean(x)) / np.std(x)

# Test Array 1
a = np.array([1, 2, 3, 4, 5])
print("Array 1:", standardize(a))

# Test Array 2
b = np.array([10, 20, 30, 40, 50])
print("Array 2:", standardize(b))

# Test Array 3
c = np.array([5, 8, 12, 15, 20])
print("Array 3:", standardize(c))
```
