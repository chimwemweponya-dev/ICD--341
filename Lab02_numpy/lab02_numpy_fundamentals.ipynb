```python
import numpy as np
```

```python
x = np.array([10, 20, 30, 40, 50])
print(x)
```

```python
print(x.shape)
print(x.ndim)
print(x.dtype)
```

- `x.shape` gives '(5,)': the size of each dimension (5 elements, one dimension).

- `x.ndim` gives '1': the number of dimensions (axes).

- `x.dtype` gives e.g. 'int64': the data type of the elements.

```python
a = np.zeros(10)
b = np.ones(10)
c = np.arange(1, 11)
d = np.linspace(0, 1, 11)

print("a:", a)
print("b", b)
print("c", c)
print("d", d)
```

- `np.zeros(10)`: array of ten 0.0s. Use: initialising an accumulator or empty results array.

- `np.ones(10)`: array of ten 1.0s. Use: a column of ones (intercept) in a regression design matrix.

 `np.arange(1, 11)`: 1 to 10, **step size** fixed, end excluded. Use: observation indices.
- `np.linspace(0, 1, 11)`: 11 **evenly spaced points** including both ends. Use: grid of x values for plotting a density.
- **Difference:** `arange` you choose the step; `linspace` you choose the number of points.

- `np.zeros()`: parameter initialisation-setting initial zero-weights for optimisation algorithms like gradient descent.
- `np.ones()`: Regression intercepts-Adding a column of 1s to a design matrix to fit the bias term(ß0).
- `np.arrange()`: discrete indexing- Generating discrete sample counts, time-series steps, or bin intervals.
- `np.linspace()`: Distribution curves- Generating smooth x-grids to evaluate and plot probability density functions


```python

marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print(marks[0])
print(marks[-1])
print(marks[:3])
print(marks[2:6])
print(marks[::2])
```

```python
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print("mean:", np.mean(marks))
print("median:", np.median(marks))
print("std:", np.std(marks))
print("min:", np.min(marks))
print("max:", np.max(marks))
print("sum:", np.sum(marks))
```

**Interpretation (adapt):** The mean mark is 70.75, and the median (70) is close to it, so the marks are fairly symmetric with no strong outliers. The standard deviation of about 14.9 means marks typically fall roughly 15 points either side of the mean. The class range is 45 to 91.
- `np.std` divides by n (population). For a sample standard deviation use `np.std(marks, ddof=1)`.

```python
data = np.array([
[45, 67, 82],
[56, 91, 73],
[64, 88, 76]
])

print(data.shape)
print(data.ndim)
print(data[0])
print(data[:, 1])
print(data[1, 1])
print(data[:2])
print(data[:, -2])
```

- `np.mean(data, axis=0)` gives **[55, 82, 77]**: mean of Mathematics, Statistics, Programming.
- `np.mean(data, axis=1)` gives **[64.67, 73.33, 76.0]**: mean per student.
- **`axis=0`** collapses the rows, calculating **down each column**. **`axis=1`** collapses the columns, calculating **across each row**.

- `axis=0` (vertical/column-wise): Operates down the rows for each column. The result collapses the rows and gives one summary value per column
- `axis=1` (horizontal/row-wise): Operates across the columns for each row. The result collapses the columns and gives one summary value per row.

```python
import numpy as np, matplotlib.pyplot as plt
x = np.random.normal(loc=50, scale=10, size=1000)
print(x.mean(), np.median(x), x.std(), x.min(), x.max())
plt.hist(x, bins=30)
plt.xlabel("Value"); plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```
